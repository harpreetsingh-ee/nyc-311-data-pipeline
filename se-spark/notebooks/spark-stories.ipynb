Story 2: Local spark Session

```python
from pathlib import Path
import time
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder.master("local[*]")
    .appName("nyc311-smoke-test")
    .getOrCreate()
)
print(f"Spark version: {spark.version}")
print(f"Spark URL: {spark.sparkContext._jsc.sc().uiWebUrl()}")
```

Story 2: Print Schema and first five rows 

```python
DATA_PATH = Path.cwd() / "data" / "311-service-requests-from-2010-to-present.csv"
start = time.perf_counter()

df = spark.read.csv(str(DATA_PATH), header=True, inferSchema=True)
print(f"Read CSV with inferSchema=True in {time.perf_counter() - start:.2f}s")

df.show(5)
print(df.columns)
df.printSchema()
```

Story 2: 
1. Count total rows
2. Group by borough and count


```python
print(df.count())
df.groupBy("Borough").count().show()
```

Story 3: without infer

```python
start = time.perf_counter()

df = spark.read.csv(str(DATA_PATH), header=True, inferSchema=False)
print(f"Read CSV with inferSchema=False in {time.perf_counter() - start:.2f}s")
```
